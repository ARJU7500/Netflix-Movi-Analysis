15. create lable for Vote_Average for categry like popular,average,below avg and not popular


15.1. user define function for categorise for Vote_Average

```python
def category(df,col,lables):

    edges=[df[col].describe()['min'],
          df[col].describe()['25%'],
          df[col].describe()['50%'],
          df[col].describe()['75%'],
          df[col].describe()['max']]
    df[col]=pd.cut(df[col],edges,labels=lables,duplicates='drop')
    return df
```

15.2 create lables and call user define function

```python
lables=['Low','below Avg','Avg','Popular']
category(df, 'Vote_Average' ,lables)
df['Vote_Average'].unique()
```
